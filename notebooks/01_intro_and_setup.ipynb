### What is Geospatial Modelling?
Geospatial modelling uses location-based data to understand patterns, risks, and solutions for public health issues. In Nigeria, this approach helps target vaccination campaigns, monitor disease outbreaks, and allocate resources more effectively.

### Nigeria Administrative Boundaries
We will use official shapefiles for Nigeria's administrative boundaries. Download the shapefile from [GADM](https://gadm.org/download_country.html) or [HDX](https://data.humdata.org/dataset/nga-administrative-boundaries) and place it in the `data/` folder. This will allow us to visualize and analyze data at the state or local government area (LGA) level.

```python
# Load and plot Nigeria's administrative boundaries
# Update the file path if your shapefile has a different name
nigeria = gpd.read_file('data/gadm41_NGA_shp/gadm41_NGA_1.shp')
fig, ax = plt.subplots(figsize=(10, 8))
nigeria.plot(ax=ax, color='lightgray', edgecolor='black')
plt.title('Nigeria - Administrative Boundaries')
plt.xlabel('Longitude')
plt.ylabel('Latitude')
plt.show()
```
