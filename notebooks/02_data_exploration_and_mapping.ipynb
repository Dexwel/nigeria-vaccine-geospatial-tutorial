# Session 2: Data Exploration & Mapping Disease Incidence in Nigeria

In this session, you will learn how to explore Nigeria-specific datasets and visualize disease incidence and health resources on a map. This is a practical step in understanding the spatial distribution of vaccine-preventable diseases.

## Objectives
- Load and preview population, health facility, and disease incidence data for Nigeria
- Visualize health facilities and disease cases on an interactive map

### Step 1: Nigeria Population Data
Download gridded population data from [WorldPop Nigeria](https://www.worldpop.org/geodata/summary?id=24777) and place it in the `data/` folder. This raster file provides population estimates at a fine spatial resolution.

### Step 2: Health Facility Locations
Download health facility data from the [Nigeria Health Facility Registry](https://hfr.health.gov.ng/) or [HDX Health Facilities](https://data.humdata.org/dataset/health-facilities-in-nigeria). Save the file as `data/health_facilities.csv` or use the provided download script.

### Step 3: Disease Incidence Data
Download measles or polio incidence data from [NCDC](https://ncdc.gov.ng/reports/weekly) or [WHO](https://www.who.int/data/gho/data/themes/topics/immunization) and save as `data/measles_cases.csv`.

### Step 4: Load and Preview Data
The following code loads the population raster, health facility locations, and disease incidence data. Adjust file paths as needed.

```
import pandas as pd
import geopandas as gpd
import rasterio

# Load population raster (WorldPop Nigeria)
pop_raster = rasterio.open('data/nga_ppp_2020.tif')

# Load health facilities (CSV)
facilities = pd.read_csv('data/health_facilities.csv')

# Load disease incidence (CSV)
disease = pd.read_csv('data/measles_cases.csv')

# Preview data
print('Health Facilities:')
print(facilities.head())
print('\nDisease Incidence:')
print(disease.head())
```

### Step 5: Map Disease Incidence and Health Facilities
We will use Folium to create an interactive map showing health facility locations. You can extend this to visualize disease cases as well.

```
import folium

# Center map on Nigeria
m = folium.Map(location=[9.082, 8.6753], zoom_start=6)

# Add health facilities as points (ensure 'lat' and 'lon' columns exist)
for idx, row in facilities.iterrows():
    folium.CircleMarker([row['lat'], row['lon']], radius=2, color='blue', fill=True, fill_opacity=0.7).add_to(m)

# Optionally, add disease cases if coordinates are available
# for idx, row in disease.iterrows():
#     folium.CircleMarker([row['lat'], row['lon']], radius=3, color='red', fill=True, fill_opacity=0.5).add_to(m)

# Save and display map
m.save('data/nigeria_health_facilities_map.html')
m
```
