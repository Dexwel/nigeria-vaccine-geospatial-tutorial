### Step 1: Predicting Future Outbreaks
You can use your trained model to predict disease risk for new data, such as projected population or vaccination coverage for the next year. This helps public health officials plan interventions.

```
# Example: Predicting future risk using the trained model
# Prepare new features for prediction (update as needed)
# For demonstration, we'll use the same features as before
future_X = merged[['population', 'facilities', 'vacc_coverage']]
future_preds = clf.predict(future_X)

# Add predictions to the GeoDataFrame
merged['future_risk'] = future_preds

# Plot predicted risk areas
fig, ax = plt.subplots(figsize=(10, 8))
merged.plot(column='future_risk', ax=ax, legend=True, cmap='coolwarm', edgecolor='black')
plt.title('Predicted Disease Risk (Future Scenario)')
plt.xlabel('Longitude')
plt.ylabel('Latitude')
plt.show()
```

### Step 2: Interactive Maps for Stakeholders
Interactive maps help communicate results to non-technical audiences. You can use Folium to create web-based maps showing predicted risk areas.

```
import folium

# Create a map centered on Nigeria
m = folium.Map(location=[9.082, 8.6753], zoom_start=6)

# Add predicted risk areas (ensure 'lat' and 'lon' columns exist or use centroids)
for idx, row in merged.iterrows():
    # Use centroid if lat/lon not available
    if hasattr(row.geometry, 'centroid'):
        lat = row.geometry.centroid.y
        lon = row.geometry.centroid.x
    else:
        lat = row['lat']
        lon = row['lon']
    color = 'red' if row['future_risk'] else 'green'
    folium.CircleMarker([lat, lon], radius=5, color=color, fill=True, fill_opacity=0.7).add_to(m)

# Save and display the map
m.save('data/nigeria_predicted_risk_map.html')
m
```
